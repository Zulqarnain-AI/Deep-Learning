```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense,Bidirectional,Input
```

```python
(x_train,y_train),(x_test,y_test) = imdb.load_data(num_words=10000)
```

```python
x_train = pad_sequences(x_train,100)
x_test = pad_sequences(x_test,100)
```

```python
model = Sequential()
model.add(Input(shape=(100,)))
model.add(Embedding(10000,32))
model.add(SimpleRNN(5))
model.add(Dense(1,activation='sigmoid'))

model.summary()

model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
history = model.fit(x_train,y_train,epochs=10,batch_size=32,validation_split=0.2)
```

```python
model1 = Sequential()
model1.add(Input(shape=(100,)))
model1.add(Embedding(10000,32))
model1.add(Bidirectional(SimpleRNN(5,return_sequences=True)))
model1.add(Bidirectional(SimpleRNN(5)))
model1.add(Dense(1,activation='sigmoid'))

model1.summary()

model1.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
history = model1.fit(x_train,y_train,epochs=10,batch_size=32,validation_split=0.2)
```
