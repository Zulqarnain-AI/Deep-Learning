```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense,Input,LSTM,GRU
```

```python
(x_train,y_train),(x_test,y_test) = imdb.load_data(num_words=10000)
x_train = pad_sequences(x_train,maxlen=100)
x_test = pad_sequences(x_test,maxlen=100)
```

```python
model = Sequential()
model.add(Input(shape=(100,))) 
model.add(Embedding(input_dim=10000, output_dim=32))
model.add(SimpleRNN(5,return_sequences=True))
model.add(SimpleRNN(5))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
history = model.fit(x_train,y_train,epochs=10,batch_size=32,validation_split=0.2)
```

```python
model2 = Sequential()
model2.add(Input(shape=(100,))) 
model2.add(Embedding(input_dim=10000, output_dim=32))
model2.add(LSTM(5,return_sequences=True))
model2.add(LSTM(5))
model2.add(Dense(1,activation='sigmoid'))

model2.summary()

model2.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
history = model2.fit(x_train,y_train,epochs=10,batch_size=32,validation_split=0.2)
```
