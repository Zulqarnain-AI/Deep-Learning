```python
from keras.datasets import imdb
from keras.models import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
from keras.src.utils import pad_sequences
```

```python
(x_train, y_train), (x_test, y_test) = imdb.load_data()
```

```python
x_train = pad_sequences(x_train, padding='post',maxlen=50)
x_test = pad_sequences(x_test, padding='post',maxlen=50)
```

```python
x_train.shape
```

```python
import numpy as np
max_index = max(np.max(x_train), np.max(x_test))
model = Sequential()
model.add(Embedding(max_index+1, 2,input_length=50))
model.add(SimpleRNN(units=32,return_sequences=False))
model.add(Dense(units=1, activation='sigmoid'))

```

```python
model.build(input_shape=(None, 50))
model.summary()
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(x_train, y_train, epochs=5,validation_data=(x_test, y_test))
```

```python
from keras.datasets import imdb
from keras.models import Sequential
from keras.layers import Dense,SimpleRNN,Embedding
from keras.src.utils import pad_sequences
```

```python
(x_train,y_train),(x_test,y_test) = imdb.load_data(maxlen=10000)
```

```python
len(x_test)
```

```python
max_word = max(x_train.max(),x_test.max())+1
x_train = pad_sequences(x_train,padding = 'post',maxlen=100)
x_test = pad_sequences(x_test,padding='post',maxlen=100)

model = Sequential()
model.add(Embedding(input_dim=max_word,output_dim=50,input_length = 100))
model.add(SimpleRNN(32,return_sequences=False))
model.add(Dense(1,activation='sigmoid'))

model.compile(optimizer='adam', loss='binary_crossentropy',metrics=['accuracy'])
model.fit(x_train,y_train,epochs=10,validation_data=(x_test,y_test))

```
